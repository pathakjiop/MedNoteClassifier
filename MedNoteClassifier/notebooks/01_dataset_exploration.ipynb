# 01 — Dataset exploration

Inspect the **actual** NBME Kaggle files. This notebook does not invent SOAP labels.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
import pandas as pd
from src.data_loader import load_raw_frames, parse_span_list, parse_annotation_list
train, notes, features, test = load_raw_frames()
print(train.shape, notes.shape, features.shape)
print('train columns', list(train.columns))
print('notes columns', list(notes.columns))
print('features columns', list(features.columns))
print(train.head(3))
print('unique notes (all)', notes.pn_num.nunique(), 'annotated', train.pn_num.nunique())
print('missing train', train.isna().sum().to_dict())
print('duplicate ids', train.id.duplicated().sum())
```

## Annotation analysis

```python
train = train.copy()
train['n_spans'] = train.location.map(lambda x: len(parse_span_list(x)))
print('empty', (train.n_spans==0).sum(), 'nonempty', (train.n_spans>0).sum())
print('annotations per note')
print(train.groupby('pn_num').n_spans.sum().describe())
print(features.groupby('case_num').size())
print('sample feature labels:')
print(features.feature_text.head(15).tolist())
```
